# Results Printing and Interpretation

This notebook uses the `.txt` files from the subgroup discovery results and prints summary statistics compared to the data set.

```python
# imports
# include additional libraries in case of plotting
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import re
```

```python
# Import full dataset
df = pd.read_csv("feature_engineered_data/df_final.csv")
```

```python
def check_results(lines):
    """Print info, flag possible false discoveries."""

    # Regex to capture conditions inside the IF statement, taking ANDs into account
    pattern = r"(IF|AND)([A-Za-z0-9_ ]+?)\s*(<=|>=|<|>|=)\s*([0-9.eE+-]+)"
    
    # go over each subgroup found
    for line in lines:
        
        # Extract all _/variable/operator/value groups
        conditions = re.findall(pattern, line.split("->")[0])
        proportions = []

        for _, var, op, val in conditions:
            var = var.strip()
            val = float(val)

            if op == "<=":
                cond = df[var] <= val
            elif op == ">=":
                cond = df[var] >= val
            elif op == "<":
                cond = df[var] < val
            elif op == ">":
                cond = df[var] > val
            elif op == "=":
                cond = df[var] == val
            else:
                continue

            prop = cond.sum() / 236
            proportions.append(prop)
            
            # Compute descriptive info
            var_mean = df[var].mean()
            var_min = df[var].min()
            var_max = df[var].max()

            # calculate descriptive info vs subgroup for printing
            if abs(val - var_mean) / abs(var_mean) < 0.01:
                relation_to_mean = "equal to mean" # relative to mean
            elif val < var_mean:
                relation_to_mean = "below mean"
            else:
                relation_to_mean = "above mean"

            if val == var_min:
                relation_to_range = "equal to min"
            elif val == var_max:
                relation_to_range = "equal to max"
            elif val < var_min:
                relation_to_range = "below min"
            elif val > var_max:
                relation_to_range = "above max"
            else:
                relation_to_range = "within range"

            print(f"{var} {op} {val:.4f} -> proportion = {prop:.3f}")
            print(f"   ==> threshold is {relation_to_mean}, {relation_to_range}")
            

        # possible false discovery flagging
        if all(p >= 0.5 for p in proportions):
            print("False discovery (all >= 50%)")
            
        print("\n")
```

## Welch's t-stat results

```python
# Read rules from file max
with open("result_files/score_tstat_max.txt", "r") as f:
    t_max_lines = [l.strip() for l in f if l.strip()]
```

```python
check_results(t_max_lines)
```
